```python
subset = df[['plntname', 'implmass', 'imangsep', 'imdist', 'impltemp']].copy()
print(subset.shape)
subset.head()
```

```python
subset = subset.dropna(subset=['implmass', 'imangsep', 'imdist'])
print(f"Rows remaining after dropping missing values: {subset.shape[0]}")
```

```python
#true_sep (AU) = angular_sep (arcsec) x distance (pc)
subset['true_sep_au'] = subset['imangsep'] * subset['imdist']

subset.head()
```

```python
fig, ax = plt.subplots(figsize=(8, 6))

ax.scatter(subset['true_sep_au'], subset['implmass'],
           color='#4C72B0', edgecolor='black', alpha=0.7, s=60)

ax.set_xlabel('True Separation (AU)')
ax.set_ylabel('Companion Mass (M$_{Jup}$)')
ax.set_title('Direct Imaging: Companion Mass vs. Separation (n=44)')
ax.set_xscale('log')

plt.tight_layout()
plt.show()
```
