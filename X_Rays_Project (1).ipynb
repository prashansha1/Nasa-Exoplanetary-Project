```python

from sunpy.net import Fido, attrs as a
from sunpy.timeseries import TimeSeries
```

```python
result = Fido.search(
    a.Time("2024-05-01", "2024-05-31"),
    a.Instrument("XRS"),
    a.goes.SatelliteNumber(16),
    a.Resolution("avg1m"),
)
print(result)
```

```python
files= Fido.fetch(result)
print(len(files), "Files downloaded")
```

```python
print(df.shape)
print(df.isna().sum())
print(df["xrsa_quality"].value_counts())
print(df["xrsb"].describe())
```

```python
import matplotlib.pyplot as plt

good = df[df["xrsb_quality"] == 0]

plt.figure(figsize=(14, 4))
plt.semilogy(good.index, good["xrsb"], linewidth=0.6)
for level, name in [(1e-6, "C"), (1e-5, "M"), (1e-4, "X")]:
    plt.axhline(level, color="gray", linestyle="--", linewidth=0.7)
    plt.text(good.index[0], level * 1.15, name, fontsize=9)
plt.ylabel("Flux (W/m²)")
plt.title("GOES XRS xrsb, May 2024")
plt.show()
```

```python
import pandas as pd

hek = Fido.search(
    a.Time("2024-05-01", "2024-05-31"),
    a.hek.EventType("FL"),
    a.hek.OBS.Observatory == "GOES",
)
flares = hek["hek"]
print(len(flares))
print(flares.colnames)
```

```python
fl = pd.DataFrame({
    "start": pd.to_datetime([str(t) for t in flares["event_starttime"]]),
    "peak":  pd.to_datetime([str(t) for t in flares["event_peaktime"]]),
    "end":   pd.to_datetime([str(t) for t in flares["event_endtime"]]),
    "class": [str(c) for c in flares["fl_goescls"]],
})
fl = fl.drop_duplicates(subset="peak").sort_values("peak").reset_index(drop=True)
print(fl.shape)
print(fl.head(10))
print(fl["class"].str[0].value_counts())
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(14, 4))
plt.semilogy(good.index, good["xrsb"], linewidth=0.6)
for p in fl["peak"]:
    plt.axvline(p, color="red", alpha=0.25, linewidth=0.8)
plt.ylabel("Flux (W/m²)")
plt.title("GOES xrsb with catalogued flare peaks")
plt.show()
```

```python
flux = df["xrsb"].where(df["xrsb_quality"] == 0)
flux = flux.asfreq("1min")
print(flux.isna().sum(), "missing minutes out of", len(flux))
print(flux.index.min(), flux.index.max())
```

```python
import numpy as np

HORIZON = 60

peak_flag = pd.Series(0, index=flux.index)
for p in fl["peak"]:
    p = p.round("1min")
    if p in peak_flag.index:
        peak_flag.loc[p] = 1

future = peak_flag[::-1].rolling(HORIZON, min_periods=1).max()[::-1].shift(-1)
print(peak_flag.sum(), "flare peaks marked")
print(future.value_counts())
```

```python
logf = np.log10(flux)

feat = pd.DataFrame(index=flux.index)
feat["log_flux"] = logf
for w in [10, 30, 60]:
    feat[f"mean_{w}"]  = logf.rolling(w).mean()
    feat[f"max_{w}"]   = logf.rolling(w).max()
    feat[f"std_{w}"]   = logf.rolling(w).std()
    feat[f"slope_{w}"] = logf - logf.shift(w)

print(feat.shape)
print(feat.tail())
```

```python
data = feat.copy()
data["label"] = future
data = data.iloc[:-HORIZON].dropna()
data = data.iloc[::10]

print(data.shape)
print(data["label"].value_counts())
print("flare fraction:", data["label"].mean())
```

```python
split = int(len(data) * 0.7)
train, test = data.iloc[:split], data.iloc[split:]

X_train, y_train = train.drop(columns="label"), train["label"]
X_test,  y_test  = test.drop(columns="label"),  test["label"]

print(train.index.min(), "to", train.index.max())
print(test.index.min(),  "to", test.index.max())
print("train flare fraction:", y_train.mean())
print("test flare fraction: ", y_test.mean())
```

```python
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix

lr = make_pipeline(
    StandardScaler(),
    LogisticRegression(class_weight="balanced", max_iter=1000)
)
lr.fit(X_train, y_train)

pred_lr = lr.predict(X_test)
print(confusion_matrix(y_test, pred_lr))
print(classification_report(y_test, pred_lr, digits=3))
```

```python
from sklearn.dummy import DummyClassifier

dummy = DummyClassifier(strategy="most_frequent").fit(X_train, y_train)
pred_d = dummy.predict(X_test)
print("accuracy:", (pred_d == y_test).mean())
print(classification_report(y_test, pred_d, digits=3, zero_division=0))
```

```python
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(
    n_estimators=300, class_weight="balanced_subsample",
    min_samples_leaf=5, n_jobs=-1, random_state=42
)
rf.fit(X_train, y_train)

pred_rf = rf.predict(X_test)
print(confusion_matrix(y_test, pred_rf))
print(classification_report(y_test, pred_rf, digits=3))
```

```python
from sklearn.metrics import PrecisionRecallDisplay, average_precision_score

fig, ax = plt.subplots(figsize=(6, 5))
for name, model in [("LogReg", lr), ("RandomForest", rf)]:
    scores = model.predict_proba(X_test)[:, 1]
    ap = average_precision_score(y_test, scores)
    PrecisionRecallDisplay.from_predictions(y_test, scores, name=f"{name} (AP={ap:.2f})", ax=ax)
ax.axhline(y_test.mean(), color="gray", linestyle="--", label="random")
ax.legend()
plt.show()
```

```python
imp = pd.Series(rf.feature_importances_, index=X_train.columns).sort_values()
imp.plot.barh(figsize=(6, 5))
plt.title("Random forest feature importance")
plt.show()
```

```python
files_y = Fido.fetch(files_y, max_conn=2, progress=False)
print(len(files_y), "downloaded")
print(len(files_y.errors), "still failing")
```

```python
ts_y = TimeSeries(files_y, concatenate=True)
df_y = ts_y.to_dataframe()
flux_y = df_y["xrsb"].where(df_y["xrsb_quality"] == 0)
flux_y = flux_y[~flux_y.index.duplicated()].asfreq("1min")
print(flux_y.index.min(), flux_y.index.max())
print(flux_y.isna().sum(), "missing minutes out of", len(flux_y))
```

```python
frames = []
for m in pd.date_range("2024-01-01", "2024-12-01", freq="MS"):
    t0 = m.strftime("%Y-%m-%d")
    t1 = (m + pd.offsets.MonthEnd(0)).strftime("%Y-%m-%d 23:59:59")
    try:
        r = Fido.search(a.Time(t0, t1), a.hek.EventType("FL"), a.hek.OBS.Observatory == "GOES")
        f = r["hek"]
        frames.append(pd.DataFrame({
            "start": pd.to_datetime([str(t) for t in f["event_starttime"]]),
            "peak":  pd.to_datetime([str(t) for t in f["event_peaktime"]]),
            "end":   pd.to_datetime([str(t) for t in f["event_endtime"]]),
            "class": [str(c) for c in f["fl_goescls"]],
        }))
        print(m.strftime("%b"), len(f))
    except Exception as e:
        print(m.strftime("%b"), "FAILED:", e)

fl_y = pd.concat(frames).drop_duplicates(subset="peak").sort_values("peak").reset_index(drop=True)
print(fl_y.shape)
print(fl_y["class"].str[0].value_counts())
```

```python
def make_dataset(flux, fl, horizon=60, step=10):
    peak_flag = pd.Series(0, index=flux.index)
    for p in fl["peak"]:
        p = p.round("1min")
        if p in peak_flag.index:
            peak_flag.loc[p] = 1
    future = peak_flag[::-1].rolling(horizon, min_periods=1).max()[::-1].shift(-1)

    logf = np.log10(flux)
    feat = pd.DataFrame(index=flux.index)
    feat["log_flux"] = logf
    for w in [10, 30, 60]:
        feat[f"mean_{w}"]  = logf.rolling(w).mean()
        feat[f"max_{w}"]   = logf.rolling(w).max()
        feat[f"std_{w}"]   = logf.rolling(w).std()
        feat[f"slope_{w}"] = logf - logf.shift(w)

    data = feat.copy()
    data["label"] = future
    data = data.iloc[:-horizon].dropna()
    return data.iloc[::step]

data_y = make_dataset(flux_y, fl_y)
print(data_y.shape)
print(data_y["label"].value_counts())
print("flare fraction:", data_y["label"].mean())

split = int(len(data_y) * 0.7)
train_y, test_y = data_y.iloc[:split], data_y.iloc[split:]
Xtr, ytr = train_y.drop(columns="label"), train_y["label"]
Xte, yte = test_y.drop(columns="label"), test_y["label"]
print("train:", train_y.index.min(), "to", train_y.index.max(), "| flare frac", ytr.mean())
print("test: ", test_y.index.min(),  "to", test_y.index.max(),  "| flare frac", yte.mean())
```

```python
from sklearn.metrics import average_precision_score

lr_y = make_pipeline(StandardScaler(), LogisticRegression(class_weight="balanced", max_iter=1000)).fit(Xtr, ytr)
rf_y = RandomForestClassifier(n_estimators=300, class_weight="balanced_subsample",
                              min_samples_leaf=5, n_jobs=-1, random_state=42).fit(Xtr, ytr)

for name, model in [("LogReg", lr_y), ("RandomForest", rf_y)]:
    print("=====", name)
    print(confusion_matrix(yte, model.predict(Xte)))
    print(classification_report(yte, model.predict(Xte), digits=3))
    print("Average precision:", round(average_precision_score(yte, model.predict_proba(Xte)[:, 1]), 3))
```

```python
fig, ax = plt.subplots(figsize=(6, 5))
for name, model in [("LogReg", lr_y), ("RandomForest", rf_y)]:
    s = model.predict_proba(Xte)[:, 1]
    PrecisionRecallDisplay.from_predictions(yte, s, name=name, ax=ax)
ax.axhline(yte.mean(), color="gray", linestyle="--", label="random")
ax.legend()
ax.set_title("Flare in next 60 min: precision-recall (2024, time-split)")
plt.savefig("pr_curve.png", dpi=150, bbox_inches="tight")
plt.show()
```

```python
for col in ["log_flux", "slope_10", "max_60", "std_10"]:
    print(col, "AP alone:", round(average_precision_score(yte, Xte[col]), 3))
print("random:", round(yte.mean(), 3))
```

```python
H = 24 * 60
fut_max = flux_y[::-1].rolling(H, min_periods=H // 2).max()[::-1].shift(-1)
y24 = (fut_max >= 1e-5).astype(float).where(fut_max.notna())

xa = df_y["xrsa"].where(df_y["xrsa_quality"] == 0)
xa = xa[~xa.index.duplicated()].asfreq("1min")
logb, loga = np.log10(flux_y), np.log10(xa)

F = pd.DataFrame(index=flux_y.index)
F["log_flux"] = logb
F["hardness"] = loga - logb
for w in [60, 360, 1440]:
    F[f"max_{w}"]  = logb.rolling(w, min_periods=w // 2).max()
    F[f"mean_{w}"] = logb.rolling(w, min_periods=w // 2).mean()
    F[f"std_{w}"]  = logb.rolling(w, min_periods=w // 2).std()
F["slope_60"] = logb - logb.shift(60)

D = F.copy()
D["label"] = y24
D = D.dropna().iloc[::30]

cut = pd.Timestamp("2024-09-01")
tr = D[D.index < cut - pd.Timedelta("1D")]   # 1-day gap so labels don't overlap the test period
te = D[D.index >= cut]
X2tr, y2tr = tr.drop(columns="label"), tr["label"]
X2te, y2te = te.drop(columns="label"), te["label"]
print(X2tr.shape, X2te.shape)
print("train M+ fraction:", y2tr.mean(), "| test M+ fraction:", y2te.mean())
```

```python
lr2 = make_pipeline(StandardScaler(), LogisticRegression(class_weight="balanced", max_iter=2000)).fit(X2tr, y2tr)
rf2 = RandomForestClassifier(n_estimators=300, class_weight="balanced_subsample",
                             min_samples_leaf=10, n_jobs=-1, random_state=42).fit(X2tr, y2tr)

print("random:", round(y2te.mean(), 3))
for col in ["log_flux", "max_1440", "max_360"]:
    print(col, "alone:", round(average_precision_score(y2te, X2te[col]), 3))
for name, m in [("LogReg", lr2), ("RandomForest", rf2)]:
    s = m.predict_proba(X2te)[:, 1]
    print(name, "AP:", round(average_precision_score(y2te, s), 3))
    print(confusion_matrix(y2te, (s >= 0.5).astype(int)))
```
